#### (i) Supervised Learning

```python
x = data.iloc[:, 2:].values
y = data.iloc[:, 1].map({'B': 0, 'M': 1}).values 

m = 30  

trainaccs = []
trainprecisions = []
trainrecalls = []
trainf1s = []
trainaucs = []

testaccs = []
testprecisions = []
testrecalls = []
testf1s = []
testaucs = []

lastrunfprtrain = None
lastruntprtrain = None
lastrunfprtest = None
lastruntprtest = None

lastruntraintrue = None
lastruntrainpred = None
lastruntesttrue = None
lastruntestpred = None

clist = [0.01, 0.1, 1.0, 10.0, 100.0]

for i in range(m):
    splitter = StratifiedShuffleSplit(
        n_splits=1,
        test_size=0.2,
        random_state=42 + i
    )
    
    for trainindex, testindex in splitter.split(x, y):
        xtrain = x[trainindex]
        xtest = x[testindex]
        ytrain = y[trainindex]
        ytest = y[testindex]

    scaler = StandardScaler()
    xtrainnorm = scaler.fit_transform(xtrain)
    xtestnorm = scaler.transform(xtest)

    baseclf = LinearSVC(
        penalty='l1',
        dual=False,
        max_iter=10000
    )
    paramgrid = {'C': clist}

    grid = GridSearchCV(
        estimator=baseclf,
        param_grid=paramgrid,
        cv=5,
        scoring='accuracy',
        n_jobs=-1
    )

    grid.fit(xtrainnorm, ytrain)
    bestclf = grid.best_estimator_


    ytrainpred = bestclf.predict(xtrainnorm)
    ytestpred = bestclf.predict(xtestnorm)

    ytrainscore = bestclf.decision_function(xtrainnorm)
    ytestscore = bestclf.decision_function(xtestnorm)

    trainacc = accuracy_score(ytrain, ytrainpred)
    trainprecision = precision_score(ytrain, ytrainpred)
    trainrecall = recall_score(ytrain, ytrainpred)
    trainf1 = f1_score(ytrain, ytrainpred)
    trainauc = roc_auc_score(ytrain, ytrainscore)

    trainaccs.append(trainacc)
    trainprecisions.append(trainprecision)
    trainrecalls.append(trainrecall)
    trainf1s.append(trainf1)
    trainaucs.append(trainauc)

    testacc = accuracy_score(ytest, ytestpred)
    testprecision = precision_score(ytest, ytestpred)
    testrecall = recall_score(ytest, ytestpred)
    testf1 = f1_score(ytest, ytestpred)
    testauc = roc_auc_score(ytest, ytestscore)

    testaccs.append(testacc)
    testprecisions.append(testprecision)
    testrecalls.append(testrecall)
    testf1s.append(testf1)
    testaucs.append(testauc)

    lastrunfprtrain, lastruntprtrain, _ = roc_curve(ytrain, ytrainscore)
    lastrunfprtest, lastruntprtest, _ = roc_curve(ytest, ytestscore)

    lastruntraintrue = ytrain
    lastruntrainpred = ytrainpred
    lastruntesttrue = ytest
    lastruntestpred = ytestpred

print('supervised l1 svm, m =', m)
print('train average accuracy:', np.mean(trainaccs))
print('train average precision:', np.mean(trainprecisions))
print('train average recall:', np.mean(trainrecalls))
print('train average f1:', np.mean(trainf1s))
print('train average auc:', np.mean(trainaucs))
print()
print('test average accuracy:', np.mean(testaccs))
print('test average precision:', np.mean(testprecisions))
print('test average recall:', np.mean(testrecalls))
print('test average f1:', np.mean(testf1s))
print('test average auc:', np.mean(testaucs))

plt.figure()
plt.plot(lastrunfprtrain, lastruntprtrain, label='train roc')
plt.plot(lastrunfprtest, lastruntprtest, label='test roc')
plt.plot([0, 1], [0, 1], linestyle='--')
plt.xlabel('false positive rate')
plt.ylabel('true positive rate')
plt.title('l1 svm roc (one run)')
plt.legend()
plt.grid(True)
plt.show()


traincm = confusion_matrix(lastruntraintrue, lastruntrainpred)
testcm = confusion_matrix(lastruntesttrue, lastruntestpred)

print('train confusion matrix one run:')
print(traincm)
print()
print('test confusion matrix one run:')
print(testcm)
```

```python
trainaccuracy = []
trainprecision = []
trainrecall = []
trainf1 = []
trainauc = []

m = 30

for i in range(m):
    xtrain, xtest, ytrain, ytest = train_test_split(x, y, test_size=0.2, stratify=y, random_state=i)
    scaler = StandardScaler()
    xtrainscaled = scaler.fit_transform(xtrain)
    kmeans = KMeans(n_clusters=2, random_state=i, n_init=20)
    kmeans.fit(xtrainscaled)
    distances= kmeans.transform(xtrainscaled)
    sortedidx0 = np.argsort(distances[:, 0])
    sortedidx1= np.argsort(distances[:, 1])
    
    top30idx0 =  sortedidx0[:30]
    top30idx1= sortedidx1[:30]
    ytrainreset = ytrain.reset_index(drop=True)
    labels0 =   ytrainreset.iloc[top30idx0]
    labels1 = ytrainreset.iloc[top30idx1]
    
    poll0 = labels0.mode()[0]
    poll1 = labels1.mode()[0]
    cluster0label =poll0
    cluster1label = poll1
    
    predlabels =np.zeros_like(kmeans.labels_)
    predlabels[kmeans.labels_== 0] = cluster0label
    predlabels[kmeans.labels_ == 1] = cluster1label
    expdist = np.exp(-distances)
    probs = expdist / expdist.sum(axis=1, keepdims=True)
    
    if cluster0label ==1:
        yprobas = probs[:, 0]
    else:
        yprobas = probs[:, 1]




    trainaccuracy.append(accuracy_score(ytrain, predlabels))
    trainprecision.append(precision_score(ytrain, predlabels))
    trainrecall.append(recall_score(ytrain, predlabels))
    trainf1.append(f1_score(ytrain, predlabels))
    trainauc.append(roc_auc_score(ytrain, yprobas))
    if i == 0:
        print("confusion matrix for training run 1:")
        print(confusion_matrix(ytrain, predlabels))
        fpr, tpr, _ = roc_curve(ytrain, yprobas)
        plt.figure()
        plt.plot(fpr, tpr, label='roc curve')
        plt.plot([0, 1], [0, 1], 'k--')
        plt.xlabel('fpr')
        plt.ylabel('tpr')
        plt.title('roc ')
        plt.legend()
        plt.show()
print(f"accuracy:{np.mean(trainaccuracy)}")
print(f"precision:{np.mean(trainprecision)}")
print(f"recall:{np.mean(trainrecall)}")
print(f"f1 score:{np.mean(trainf1)}")
print(f"auc:{np.mean(trainauc)}")
```

#### (iv) Spectral Clustering

```python
from sklearn.cluster import SpectralClustering
trainaccuracy = []
trainprecision = []
trainrecall = []
trainf1 = []
trainauc = []

testaccuracy = []
testprecision = []
testrecall = []
testf1 = []
testauc = []

m = 30

for i in range(m):
    xtrain, xtest, ytrain, ytest = train_test_split(x, y, test_size=0.2, stratify=y, random_state=i)
    
    ss = StandardScaler()
    xtrainscaled = ss.fit_transform(xtrain)
    xtestscaled = ss.transform(xtest)
    
    model = SpectralClustering(n_clusters=2, affinity='rbf', gamma=1, random_state=i)
    lbs = model.fit_predict(xtrainscaled)
    
    idx0 = np.where(lbs == 0)[0]
    idx1 = np.where(lbs == 1)[0]
    
    ytrv = ytrain.values
    
    if len(idx0) > 0:
        c0lab = pd.Series(ytrv[idx0]).mode()[0]
    else:
        c0lab = 0
        
    if len(idx1) > 0:
        c1lab = pd.Series(ytrv[idx1]).mode()[0]
    else:
        c1lab = 1
        
    finalpredtr = np.zeros_like(lbs)
    finalpredtr[lbs == 0] = c0lab
    finalpredtr[lbs == 1] = c1lab
    
    trainaccuracy.append(accuracy_score(ytrain, finalpredtr))
    trainprecision.append(precision_score(ytrain, finalpredtr))
    trainrecall.append(recall_score(ytrain, finalpredtr))
    trainf1.append(f1_score(ytrain, finalpredtr))
    trainauc.append(roc_auc_score(ytrain, finalpredtr))

    modeltest = SpectralClustering(n_clusters=2, affinity='rbf', gamma=1, random_state=i)
    lbstest = modeltest.fit_predict(xtestscaled)
    
    idx0te = np.where(lbstest == 0)[0]
    idx1te = np.where(lbstest == 1)[0]
    
    ytev = ytest.values
    if len(idx0te) > 0:
        c0labte = pd.Series(ytev[idx0te]).mode()[0]
    else:
        c0labte = 0
    if len(idx1te) > 0:
        c1labte = pd.Series(ytev[idx1te]).mode()[0]
    else:
        c1labte = 1
        
    finalpredte = np.zeros_like(lbstest)
    finalpredte[lbstest == 0] = c0labte
    finalpredte[lbstest == 1] = c1labte
    
    testaccuracy.append(accuracy_score(ytest, finalpredte))
    testprecision.append(precision_score(ytest, finalpredte))
    testrecall.append(recall_score(ytest, finalpredte))
    testf1.append(f1_score(ytest, finalpredte))
    testauc.append(roc_auc_score(ytest, finalpredte))
    
    if i == 0:
        print("confusion matrix train run 1")
        print(confusion_matrix(ytrain, finalpredtr))
        
        fpr, tpr, _ = roc_curve(ytrain, finalpredtr)
        plt.figure()
        plt.plot(fpr, tpr)
        plt.title('roc train')
        plt.show()

        print("confusion matrix test run 1")
        print(confusion_matrix(ytest, finalpredte))
        
        fprte, tprte, _ = roc_curve(ytest, finalpredte)
        plt.figure()
        plt.plot(fprte, tprte)
        plt.title('roc test')
        plt.show()

print("avg train acc")
print(np.mean(trainaccuracy))
print("avg train pre")
print(np.mean(trainprecision))
print("avg train rec")
print(np.mean(trainrecall))
print("avg train f1")
print(np.mean(trainf1))
print("avg train auc")
print(np.mean(trainauc))

print("avg test acc")
print(np.mean(testaccuracy))
print("avg test pre")
print(np.mean(testprecision))
print("avg test rec")
print(np.mean(testrecall))
print("avg test f1")
print(np.mean(testf1))
print("avg test auc")
print(np.mean(testauc))
```

### (b) Repeat each of the following two procedures 50 times

#### (i) Passive

#### (ii) Active

```python
allerr = []
columns = ['a1','a2','a3','class']
data = pd.read_csv('../Data/data_banknote_authentication.txt',header=None,names=columns)
x = data.iloc[:,:-1]
y = data.iloc[:,-1]
for i in range(50):
    xtrain, xtest, ytrain, ytest = train_test_split(x, y, test_size=472, random_state=i)
    xtrain = xtrain.reset_index(drop=True)
    ytrain = ytrain.reset_index(drop=True)
    
    idx = np.arange(len(xtrain))
    np.random.shuffle(idx)
    
    pool = list(idx[:10])
    rest = list(idx[10:])
    
    rowerr = []
    
    for j in range(90):
        poolx = xtrain.iloc[pool]
        pooly = ytrain.iloc[pool]
        
        param = {'C': [0.1, 1, 10, 100, 1000]}
        svm = LinearSVC(penalty='l1', dual=False, max_iter=5000)
        clf = GridSearchCV(svm, param, cv=5, n_jobs=-1)
        
        try:
            clf.fit(poolx, pooly)
            ypred = clf.predict(xtest)
            acc = accuracy_score(ytest, ypred)
            err = 1 - acc
            rowerr.append(err)
            
            if j < 89:
                restx = xtrain.iloc[rest]
                dist = np.abs(clf.decision_function(restx))
                sortedidx = np.argsort(dist)
                top10 = sortedidx[:10]
                
                toadd = []
                for k in top10:
                    toadd.append(rest[k])
                
                for val in toadd:
                    pool.append(val)
                    rest.remove(val)
        except:
            rowerr.append(1.0)
            if j < 89:
                toadd = rest[:10]
                for val in toadd:
                    pool.append(val)
                    rest.remove(val)
                    
    allerr.append(rowerr)

finalerrors1 = pd.DataFrame(allerr)
print(finalerrors1)
```
